```python
import numpy as np
import matplotlib.pyplot as plt

# Load DOS data
dos_data = np.loadtxt('co3o4.dos', comments='#')
energy = dos_data[:, 0]
dos = dos_data[:, 1]

# Get Fermi energy from output file
fermi = None
with open('co3o4.out', 'r') as f: # Changed from 'co3o4_dos.out' to 'co3o4.out'
    for line in f:
        if 'Fermi' in line:
            fermi = float(line.split()[-2])
            break

print(f"Fermi energy: {fermi} eV")

# Plot
plt.figure(figsize=(8, 5))
plt.plot(energy - fermi, dos, color='navy', linewidth=1.5)
plt.axvline(x=0, color='red', linestyle='--', linewidth=1, label='Fermi level')
plt.xlabel('Energy (eV)', fontsize=13)
plt.ylabel('DOS (states/eV)', fontsize=13)
plt.title('Density of States — Co₃O₄', fontsize=14)
plt.legend(fontsize=11)
plt.xlim(-10, 10)
plt.ylim(bottom=0)
plt.tight_layout()
plt.savefig('co3o4_dos.png', dpi=300)
plt.show()
print("DOS plot saved as co3o4_dos.png")
```

```python
from scipy.ndimage import gaussian_filter1d

dos_smooth = gaussian_filter1d(dos, sigma=5)

plt.figure(figsize=(8, 5))
plt.plot(energy - fermi, dos_smooth, color='navy', linewidth=2)
plt.axvline(x=0, color='red', linestyle='--', linewidth=1, label='Fermi level')
plt.fill_between(energy - fermi, dos_smooth, alpha=0.2, color='navy')
plt.xlabel('Energy (eV)', fontsize=13)
plt.ylabel('DOS (states/eV)', fontsize=13)
plt.title('Density of States — Co₃O₄', fontsize=14)
plt.legend(fontsize=11)
plt.xlim(-10, 10)
plt.ylim(bottom=0)
plt.tight_layout()
plt.savefig('co3o4_dos_smooth.png', dpi=300)
plt.show()
```
