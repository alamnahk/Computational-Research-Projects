```python
import os

yamuna_data_path = '/content/data/yamuna'

# Get the size of the directory in human-readable format
def get_dir_size(path):
    total_size = 0
    for dirpath, dirnames, filenames in os.walk(path):
        for f in filenames:
            fp = os.path.join(dirpath, f)
            if not os.path.islink(fp):
                total_size += os.path.getsize(fp)
    # Convert bytes to megabytes, gigabytes etc.
    for unit in ['bytes', 'KB', 'MB', 'GB', 'TB']:
        if total_size < 1024.0:
            return f"{total_size:3.1f} {unit}"
        total_size /= 1024.0
    return f"{total_size:3.1f} PB"

size = get_dir_size(yamuna_data_path)
print(f"The Yamuna samples in '{yamuna_data_path}' currently occupy: {size}")

```

### Verifying Integrity of Downloaded Hospital Samples with FastQC

Given the `disk-limit exceeded` error during download, it's possible that some files are incomplete. We will use FastQC to re-evaluate the quality and integrity of the *successfully downloaded* hospital samples.

```python
import os
import shutil

hospital_data_path_drive = '/content/drive/MyDrive/AMR_Project/data/hospital'
local_fastqc_input_path = '/content/hospital_samples_for_fastqc_check'
local_fastqc_output_path = '/content/qc/hospital_integrity_check'

# Samples identified as fully downloaded in the previous step
fully_downloaded_samples = ['SRR13227006', 'SRR13227007', 'SRR13227008']

# Create local directories
os.makedirs(local_fastqc_input_path, exist_ok=True)
os.makedirs(local_fastqc_output_path, exist_ok=True)

print(f"Copying {len(fully_downloaded_samples)} samples from Google Drive to local storage...")
for acc in fully_downloaded_samples:
    for suffix in ['_1.fastq', '_2.fastq']:
        src_path = os.path.join(hospital_data_path_drive, f"{acc}{suffix}")
        dst_path = os.path.join(local_fastqc_input_path, f"{acc}{suffix}")
        if os.path.exists(src_path):
            shutil.copy(src_path, dst_path)
            print(f"Copied: {os.path.basename(src_path)}")
        else:
            print(f"Warning: {src_path} not found on Google Drive. Skipping copy.")

print("Copying complete. Running FastQC on local copies...")
```

### Re-downloading missing hospital and Yamuna samples to Google Drive

```python
import os
import shutil

hospital_data_path = '/content/drive/MyDrive/AMR_Project/data/hospital'
yamuna_data_path = '/content/drive/MyDrive/AMR_Project/data/yamuna'

missing_hospital_samples = ['SRR13227003', 'SRR13227004', 'SRR13227005']
missing_yamuna_samples = ['SRR8933540'] # Identified from previous DIAMOND error

def check_and_redownload(sample_list, target_drive_path):
    os.makedirs(target_drive_path, exist_ok=True)
    local_temp_path = '/content/temp_sra_downloads'
    os.makedirs(local_temp_path, exist_ok=True)

    # Ensure we are in a neutral directory or the local temp path for fasterq-dump
    os.chdir(local_temp_path)

    redownload_candidates = []
    for acc in sample_list:
        file1_drive_exists = os.path.exists(os.path.join(target_drive_path, f"{acc}_1.fastq"))
        file2_drive_exists = os.path.exists(os.path.join(target_drive_path, f"{acc}_2.fastq"))

        if not file1_drive_exists or not file2_drive_exists:
            redownload_candidates.append(acc)

    if redownload_candidates:
        print(f"Attempting to re-download missing samples to local temp directory: {local_temp_path}")
        for acc in redownload_candidates:
            print(f"Downloading {acc}...")
            # Download to local temporary path
            !fasterq-dump {acc} --split-files --progress

            # Check if files were created locally
            local_file1 = os.path.join(local_temp_path, f"{acc}_1.fastq")
            local_file2 = os.path.join(local_temp_path, f"{acc}_2.fastq")

            if os.path.exists(local_file1) and os.path.exists(local_file2):
                print(f"Successfully downloaded {acc} locally. Copying to Drive...")
                shutil.move(local_file1, os.path.join(target_drive_path, f"{acc}_1.fastq"))
                shutil.move(local_file2, os.path.join(target_drive_path, f"{acc}_2.fastq"))
                print(f"Done copying {acc} to Drive.")
            else:
                print(f"Error: Failed to download {acc} to local temp directory. Files missing after download.")

        print("Re-download and copy attempt complete.")
    else:
        print(f"All specified samples already present in {target_drive_path}.")

    # Clean up local temporary directory
    if os.path.exists(local_temp_path):
        shutil.rmtree(local_temp_path)

print("Checking and re-downloading hospital samples...")
check_and_redownload(missing_hospital_samples, hospital_data_path)

print("\nChecking and re-downloading Yamuna samples...")
check_and_redownload(missing_yamuna_samples, yamuna_data_path)

print("\nRe-checking status of all hospital samples:")
for acc in missing_hospital_samples:
    file1_exists = os.path.exists(os.path.join(hospital_data_path, f"{acc}_1.fastq"))
    file2_exists = os.path.exists(os.path.join(hospital_data_path, f"{acc}_2.fastq"))
    if file1_exists and file2_exists:
        print(f"  {acc}: Both files present.")
    else:
        print(f"  {acc}: Still missing files (1.fastq: {file1_exists}, 2.fastq: {file2_exists}).")

print("Re-checking status of all Yamuna samples:")
for acc in missing_yamuna_samples:
    file1_exists = os.path.exists(os.path.join(yamuna_data_path, f"{acc}_1.fastq"))
    file2_exists = os.path.exists(os.path.join(yamuna_data_path, f"{acc}_2.fastq"))
    if file1_exists and file2_exists:
        print(f"  {acc}: Both files present.")
    else:
        print(f"  {acc}: Still missing files (1.fastq: {file1_exists}, 2.fastq: {file2_exists}).")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

data = {
    'Sample': ['SRR13227003', 'SRR13227004', 'SRR13227005', 'SRR13227006', 'SRR13227007', 'SRR13227008',
               'SRR8933535', 'SRR8933536', 'SRR8933537', 'SRR8933538', 'SRR8933539', 'SRR8933540'],
    'Source': ['Hospital']*6 + ['Yamuna']*6,
    'ARG_Hits': [5031, 73033, 62809, 101, 213, 20409,
                 9, 11, 6, 8, 5, 748]
}

df = pd.DataFrame(data)

hospital = df[df['Source']=='Hospital']['ARG_Hits']
yamuna = df[df['Source']=='Yamuna']['ARG_Hits']
stat, pvalue = stats.mannwhitneyu(hospital, yamuna, alternative='greater')

print("=== STATISTICAL RESULTS ===")
print(f"Hospital mean ARG hits: {hospital.mean():.1f}")
print(f"Yamuna mean ARG hits: {yamuna.mean():.1f}")
print(f"Hospital median ARG hits: {hospital.median():.1f}")
print(f"Yamuna median ARG hits: {yamuna.median():.1f}")
print(f"Mann-Whitney U statistic: {stat}")
print(f"P-value: {pvalue:.6f}")
print(f"Significant (p<0.05): {pvalue < 0.05}")

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].boxplot([hospital.values, yamuna.values], labels=['Hospital\nWastewater', 'Yamuna\nRiver'])
axes[0].set_ylabel('ARG Hits')
axes[0].set_title('ARG Abundance: Hospital vs Yamuna')
axes[0].set_yscale('log')

colors = ['#d62728']*6 + ['#1f77b4']*6
axes[1].bar(df['Sample'], df['ARG_Hits'], color=colors)
axes[1].set_yscale('log')
axes[1].set_ylabel('ARG Hits (log scale)')
axes[1].set_title('ARG Hits Per Sample')
axes[1].tick_params(axis='x', rotation=90)

plt.tight_layout()
plt.savefig('/content/drive/MyDrive/AMR_Project/ARG_comparison.png', dpi=300)
plt.show()
print("Figure saved to Drive")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Parse all result files
def parse_diamond(filepath, sample_name):
    hits = {}
    try:
        with open(filepath) as f:
            for line in f:
                parts = line.strip().split('\t')
                if len(parts) >= 2:
                    gene = parts[1].split('|')[-1] if '|' in parts[1] else parts[1]
                    hits[gene] = hits.get(gene, 0) + 1
    except:
        pass
    return pd.Series(hits, name=sample_name)

samples = {}

# Hospital
for acc in ['SRR13227003','SRR13227004','SRR13227005','SRR13227006','SRR13227007','SRR13227008']:
    path = f'/content/drive/MyDrive/AMR_Project/arg_results/hospital/{acc}.txt'
    samples[acc] = parse_diamond(path, acc)

# Yamuna
for acc in ['SRR8933535','SRR8933536','SRR8933537','SRR8933538','SRR8933539','SRR8933540']:
    path = f'/content/drive/MyDrive/AMR_Project/arg_results/yamuna/{acc}.txt'
    samples[acc] = parse_diamond(path, acc)

df = pd.DataFrame(samples).fillna(0)

# Top 25 genes overall
top_genes = df.sum(axis=1).nlargest(25).index
df_top = df.loc[top_genes]

# Heatmap
plt.figure(figsize=(14, 10))
sns.heatmap(df_top, cmap='YlOrRd', linewidths=0.5,
            xticklabels=True, yticklabels=True)
plt.title('Top 25 ARGs Across Hospital and Yamuna Samples', fontsize=14)
plt.xlabel('Sample')
plt.ylabel('Resistance Gene')
plt.tight_layout()
plt.savefig('/content/drive/MyDrive/AMR_Project/ARG_heatmap.png', dpi=300)
plt.show()
print("Heatmap saved to Drive")
```

```python
# Reads per sample (from our earlier download output)
read_counts = {
    'SRR13227003': 45176558, 'SRR13227004': 45236354,
    'SRR13227005': 54060176, 'SRR13227006': 13356468,
    'SRR13227007': 10894806, 'SRR13227008': 2433016,
    'SRR8933535': 16778296, 'SRR8933536': 16971936,
    'SRR8933537': 15642650, 'SRR8933538': 16927336,
    'SRR8933539': 14437530, 'SRR8933540': 28656696
}

df_norm = pd.DataFrame({
    'Sample': list(read_counts.keys()),
    'Source': ['Hospital']*6 + ['Yamuna']*6,
    'ARG_Hits': [5031, 73033, 62809, 101, 213, 20409,
                 9, 11, 6, 8, 5, 748],
    'Total_Reads': list(read_counts.values())
})

df_norm['ARG_per_million'] = (df_norm['ARG_Hits'] / df_norm['Total_Reads']) * 1e6

print("=== NORMALIZED ARG ABUNDANCE (hits per million reads) ===")
print(df_norm[['Sample','Source','ARG_Hits','ARG_per_million']].to_string(index=False))

hospital_norm = df_norm[df_norm['Source']=='Hospital']['ARG_per_million']
yamuna_norm = df_norm[df_norm['Source']=='Yamuna']['ARG_per_million']
stat, pvalue = stats.mannwhitneyu(hospital_norm, yamuna_norm, alternative='greater')

print(f"\nHospital mean (normalized): {hospital_norm.mean():.2f} ARGs/million reads")
print(f"Yamuna mean (normalized): {yamuna_norm.mean():.2f} ARGs/million reads")
print(f"P-value: {pvalue:.6f}")

# Save normalized data
df_norm.to_csv('/content/drive/MyDrive/AMR_Project/normalized_ARG_results.csv', index=False)
print("Saved to Drive")
```
