```python
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import requests # Added import

# Re-fetch dependencies if they might be missing due to session state or out-of-order execution
# These variables (nano_set, mdr_set) should have been defined in BYHbpBUr60MG
# If a full kernel restart occurred, these might need to be re-run from that cell as well.
# For now, we assume nano_set and mdr_set are available, and only re-fetch 'interactions'.

# Get PPI data from STRING database (copied from cell _0NaA2P96_JI)
# Assuming nano_set and mdr_set are available from previous cells.
all_targets = list(nano_set.union(mdr_set))

url = "https://string-db.org/api/json/network"
params = {
    'identifiers': '%0d'.join(all_targets),
    'species': 9606,
    'required_score': 400,
    'caller_identity': 'network_pharmacology_study'
}

r = requests.get(url, params=params, timeout=30)
interactions = r.json()

# Build network
G = nx.Graph()

for interaction in interactions:
    gene_a = interaction['preferredName_A']
    gene_b = interaction['preferredName_B']
    score = interaction['score']
    G.add_edge(gene_a, gene_b, weight=score)

print(f"Network nodes: {G.number_of_nodes()}")
print(f"Network edges: {G.number_of_edges()}")

# Calculate node centrality
degree_cent = nx.degree_centrality(G)
betweenness_cent = nx.betweenness_centrality(G)

# Top 10 hub genes by degree centrality
top_hubs = sorted(degree_cent.items(), key=lambda x: x[1], reverse=True)[:10]
print("\nTop 10 Hub Genes (Degree Centrality):")
for gene, score in top_hubs:
    print(f"  {gene}: {score:.4f}")

# Define node colors by category
overlap_targets = {'IL1B', 'TNF', 'STAT3', 'IL6', 'NFKB1'}
ag_only = ag_set - au_set - mdr_set
au_only = au_set - ag_set - mdr_set
mdr_only = mdr_set - nano_set
ag_au_shared = ag_set.intersection(au_set) - mdr_set

def node_color(node):
    if node in overlap_targets:
        return '#d62728'  # red — key overlap targets
    elif node in ag_au_shared:
        return '#ff7f0e'  # orange — shared by both nanoparticles
    elif node in ag_set:
        return '#1f77b4'  # blue — Ag only
    elif node in au_set:
        return '#ffd700'  # gold — Au only
    elif node in mdr_set:
        return '#2ca02c'  # green — MDR only
    else:
        return '#aec7e8'  # light blue — other

node_colors = [node_color(n) for n in G.nodes()]
node_sizes = [3000 * degree_cent[n] + 200 for n in G.nodes()]

# Plot
plt.figure(figsize=(18, 14))
pos = nx.spring_layout(G, k=2, seed=42)

nx.draw_networkx_edges(G, pos, alpha=0.2, edge_color='gray', width=0.8)
nx.draw_networkx_nodes(G, pos, node_color=node_colors, node_size=node_sizes, alpha=0.9)
nx.draw_networkx_labels(G, pos, font_size=7, font_weight='bold')

# Legend
legend_elements = [
    mpatches.Patch(color='#d62728', label='Key Overlap Targets (Nano + MDR)'),
    mpatches.Patch(color='#ff7f0e', label='Shared Ag-Au Targets'),
    mpatches.Patch(color='#1f77b4', label='Ag Nanoparticle Targets'),
    mpatches.Patch(color='#ffd700', label='Au Nanoparticle Targets'),
    mpatches.Patch(color='#2ca02c', label='MDR Host Targets'),
]
plt.legend(handles=legend_elements, loc='upper left', fontsize=10)
plt.title('Protein-Protein Interaction Network\nAg-Au Nanoparticles vs MDR Bacterial Targets', fontsize=14, fontweight='bold')
plt.axis('off')
plt.tight_layout()

import os
output_dir = '/content/drive/MyDrive/AMR_Project/'
os.makedirs(output_dir, exist_ok=True)
plt.savefig(os.path.join(output_dir, 'network_ppi.png'), dpi=300, bbox_inches='tight')
plt.show()
print("Network figure saved")
```

```python
import requests
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Centrality statistics table
centrality_df = pd.DataFrame({
    'Gene': list(G.nodes()),
    'Degree': [G.degree(n) for n in G.nodes()],
    'Degree_Centrality': [degree_cent[n] for n in G.nodes()],
    'Betweenness_Centrality': [betweenness_cent[n] for n in G.nodes()]
}).sort_values('Degree_Centrality', ascending=False)

print("=== TOP 15 HUB GENES BY DEGREE CENTRALITY ===")
print(centrality_df.head(15).to_string(index=False))

# Pathway enrichment via Enrichr API
def enrichr_query(gene_list, library='KEGG_2021_Human'):
    # Add gene list
    genes_str = '\n'.join(gene_list)
    payload = {'list': (None, genes_str), 'description': (None, 'nanoparticle_targets')}
    response = requests.post('https://maayanlab.cloud/Enrichr/addList', files=payload)
    user_list_id = response.json()['userListId']

    # Get enrichment results
    url = f'https://maayanlab.cloud/Enrichr/enrich?userListId={user_list_id}&backgroundType={library}'
    response = requests.get(url)
    results = response.json()[library]

    df = pd.DataFrame(results, columns=[
        'Rank', 'Term', 'P-value', 'Z-score', 'Combined_score',
        'Genes', 'Adjusted_P-value', 'Old_P-value', 'Old_Adjusted_P-value'
    ])
    return df[['Term', 'P-value', 'Adjusted_P-value', 'Combined_score', 'Genes']].head(20)

# Run enrichment on overlap targets + top hub genes
query_genes = list(nano_set.union(mdr_set))
print("\nRunning KEGG pathway enrichment...")
kegg_df = enrichr_query(query_genes, 'KEGG_2021_Human')
print("\n=== TOP 15 KEGG PATHWAYS ===")
print(kegg_df.head(15)[['Term', 'P-value', 'Adjusted_P-value', 'Combined_score']].to_string(index=False))

# Save results
centrality_df.to_csv('/content/drive/MyDrive/AMR_Project/centrality_results.csv', index=False)
kegg_df.to_csv('/content/drive/MyDrive/AMR_Project/kegg_pathways.csv', index=False)
print("\nResults saved to Drive")
```

```python
# Bubble plot of top pathways
top_pathways = kegg_df.head(15).copy()
top_pathways['neg_log_p'] = -pd.Series(top_pathways['P-value'].values.astype(float)).apply(lambda x: __import__('math').log10(x))
top_pathways['Term_short'] = top_pathways['Term'].str[:45]
top_pathways['Gene_count'] = top_pathways['Genes'].apply(lambda x: len(x) if isinstance(x, list) else len(str(x).split(',')))

fig, ax = plt.subplots(figsize=(12, 9))

scatter = ax.scatter(
    top_pathways['neg_log_p'],
    range(len(top_pathways)),
    s=top_pathways['Gene_count'] * 40,
    c=top_pathways['neg_log_p'],
    cmap='YlOrRd',
    alpha=0.85,
    edgecolors='black',
    linewidths=0.5
)

ax.set_yticks(range(len(top_pathways)))
ax.set_yticklabels(top_pathways['Term_short'], fontsize=10)
ax.set_xlabel('-log10(P-value)', fontsize=12)
ax.set_title('KEGG Pathway Enrichment Analysis\nAg-Au Nanoparticle and MDR Bacterial Targets', fontsize=13, fontweight='bold')
ax.axvline(x=-__import__('math').log10(0.05), color='gray', linestyle='--', alpha=0.5, label='p=0.05')

plt.colorbar(scatter, ax=ax, label='-log10(P-value)')
plt.tight_layout()
plt.savefig('/content/drive/MyDrive/AMR_Project/pathway_enrichment.png', dpi=300, bbox_inches='tight')
plt.show()
print("Pathway enrichment figure saved")
```
